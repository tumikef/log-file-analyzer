```python
from IPython.core.display import clear_output
log_lines = [
    "2026-09-18 10:15:32 INFO Server started",
    "2026-09-18 10:16:01 WARNING High memory usage",
    "2026-09-18 10:16:45 ERROR Database connection failed",
    "2026-09-18 10:17:02 INFO User logged in",
    "2026-09-18 10:17:30 ERROR Timeout connecting to API",
    "2026-09-18 10:18:00 INFO Request processed",
    "2026-09-18 10:18:45 WARNING Disk space low",
]

#write server.log
with open("log.txt", "w") as file:
    for line in log_lines:
        file.write(line + "\n")

# Reads server.log back in
with open("log.txt", "r") as file:
    for line in file:
        print(line.strip())

# Counts how many lines are INFO, how many are WARNING, and how many are ERROR from the log file
info_count = (len([line for line in log_lines if "INFO" in line]))
print(f"INFO: {info_count}\n")
warn_count = (len([line for line in log_lines if "WARNING" in line]))
print(f"WARNING: {warn_count}\n")
err_count = (len([line for line in log_lines if "ERROR" in line]))
print(f"ERROR: {err_count}\n")

# only the ERROR lines
for line in log_lines:
    if "ERROR" in line:
         print(line)

#write a log summary
with open("log_summary.txt", "w") as file:
            file.write(f"INFO : {info_count} \n")
            file.write(f"WARNIG: {warn_count} \n")
            file.write(f"ERROR: {err_count} \n")

#CHECK IF ANYTHING HAS BEEN WIRTTEN TO THE FILE
with open("log_summary.txt", "r") as file:
    for line in file:
     #clear_output
      print(line.strip())

```
